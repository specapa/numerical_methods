# 1.5. QR-разложение

Реализация: `lab1/qr.py` (модифицированный Грама–Шмидта + QR-алгоритм). Данные из файлов.

```python
import sys
from pathlib import Path

ROOT = Path.cwd()
if ROOT.name == "notebooks":
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT))
DATA = ROOT / "data"

from lab1.dataio import read_matrix, read_vector, read_scalar, read_tridiag
from lab1.matops import (
    mat_mul, mat_vec, identity, fmt_mat, fmt_vec, mat_sub, max_abs_mat, diag_from, transpose,
)

from lab1.qr import qr_decomp, qr_eigenvalues

A = read_matrix(DATA / "1_5" / "A.txt")
eps = read_scalar(DATA / "1_5" / "eps.txt")
```

## Исходная матрица A

```python
print("A")
print(fmt_mat(A))
print("n =", len(A))
print("ε =", eps)
```

## Матрицы Q и R

```python
Q, R = qr_decomp(A)
print("Q")
print(fmt_mat(Q))
print()
print("R")
print(fmt_mat(R))
```

## Q · R

```python
QR = mat_mul(Q, R)
print("Q · R")
print(fmt_mat(QR))
print()
print("макс. |Q R - A| =", max_abs_mat(mat_sub(QR, A)))
print("макс. |Q^T Q - E| =", max_abs_mat(mat_sub(mat_mul(transpose(Q), Q), identity(len(A)))))
```

## Собственные значения (QR-алгоритм)

```python
eigs, T, nit = qr_eigenvalues(A, eps)
print("итераций QR-алгоритма:", nit)
print("собственные значения:")
for ev in eigs:
    print(" ", ev)
```

## Сравнение с NumPy (только проверка)

```python
import numpy as np

print("numpy eigenvalues =", np.linalg.eigvals(np.array(A)))
```

## Собственный пример с комплексно-сопряжённой парой

Матрица в `data/1_5/A_complex.txt`. Здесь NumPy разрешён для нахождения и проверки комплексно-сопряжённой пары.

```python
Ac = read_matrix(DATA / "1_5" / "A_complex.txt")
print("A")
print(fmt_mat(Ac))
print()
Qc, Rc = qr_decomp(Ac)
print("Q")
print(fmt_mat(Qc))
print()
print("R")
print(fmt_mat(Rc))
print()
print("макс. |Q R - A| =", max_abs_mat(mat_sub(mat_mul(Qc, Rc), Ac)))
print()
eigs_c, Tc, nit_c = qr_eigenvalues(Ac, eps)
print("итераций:", nit_c)
print("собственные значения (наш QR-алгоритм):")
for ev in eigs_c:
    print(" ", ev)
print()
np_eigs = np.linalg.eigvals(np.array(Ac))
print("numpy.linalg.eigvals:")
print(np_eigs)
print()
print("матрица после QR-итераций (вещественная форма Шура):")
print(fmt_mat(Tc))
```
