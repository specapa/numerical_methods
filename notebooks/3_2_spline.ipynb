# 3.2. Кубический сплайн

Реализация: `lab3/spline.py`. Естественный сплайн S''(x0)=S''(xn)=0. Данные из файлов.

```python
import sys
from pathlib import Path

ROOT = Path.cwd()
if ROOT.name == "notebooks":
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT))
DATA = ROOT / "data" / "lab3"

from lab3.dataio import read_vector, read_scalar
from lab3.spline import natural_cubic_spline, spline_eval

xs = read_vector(DATA / "3_2" / "x.txt")
ys = read_vector(DATA / "3_2" / "y.txt")
x_star = read_scalar(DATA / "3_2" / "x_star.txt")
print("x =", xs)
print("y =", ys)
print("x* =", x_star)

sp = natural_cubic_spline(xs, ys)
print("a =", sp["a"])
print("b =", sp["b"])
print("c =", sp["c"])
print("d =", sp["d"])

S_star = spline_eval(sp, x_star)
print("S(x*) =", S_star)
```
