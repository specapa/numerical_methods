# 1.4. Метод вращений Якоби

Реализация: `lab1/jacobi.py`. Матрица должна быть симметричной. A и ε из файлов.

```python
import sys
from pathlib import Path

ROOT = Path.cwd()
if ROOT.name == "notebooks":
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT))
DATA = ROOT / "data"

from lab1.dataio import read_matrix, read_vector, read_scalar, read_tridiag
from lab1.matops import (
    mat_mul, mat_vec, identity, fmt_mat, fmt_vec, mat_sub, max_abs_mat, diag_from, transpose,
)

from lab1.jacobi import jacobi_eigen

A = read_matrix(DATA / "1_4" / "A.txt")
eps = read_scalar(DATA / "1_4" / "eps.txt")
print("A")
print(fmt_mat(A))
print("n =", len(A))
```

## Заданная точность

```python
print("ε =", eps)
```

## Собственные значения и векторы

```python
vals, V, nrot = jacobi_eigen(A, eps)
print("число вращений:", nrot)
print("собственные значения:", fmt_vec(vals))
print()
print("матрица собственных векторов V (по столбцам)")
print(fmt_mat(V))
print()
for j in range(len(vals)):
    vj = [V[i][j] for i in range(len(A))]
    print("λ_%d = %s" % (j + 1, fmt_vec([vals[j]])[1:-1]))
    print("v_%d =" % (j + 1), fmt_vec(vj))
```

## Проверка A · V = V · Λ

```python
Lam = diag_from(vals)
AV = mat_mul(A, V)
VL = mat_mul(V, Lam)
print("A · V")
print(fmt_mat(AV))
print()
print("V · Λ")
print(fmt_mat(VL))
print()
print("макс. |A V - V Λ| =", max_abs_mat(mat_sub(AV, VL)))
```
