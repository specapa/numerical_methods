## Проверка (трёхдиагональное умножение, без полной матрицы)

```python
Ax = [0.0] * n
Ax[0] = b[0] * x[0] + c[0] * x[1]
for i in range(1, n - 1):
    Ax[i] = a[i - 1] * x[i - 1] + b[i] * x[i] + c[i] * x[i + 1]
Ax[n - 1] = a[n - 2] * x[n - 2] + b[n - 1] * x[n - 1]
print("A x =", fmt_vec(Ax))
print("d   =", fmt_vec(d))
```
