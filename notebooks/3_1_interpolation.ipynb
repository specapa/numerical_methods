# 3.1. Интерполяция Лагранжа и Ньютона

Реализация: `lab3/interpolation.py`. Узлы и X* читаются из файлов; значения функции `arcctg(x)` вычисляются по узлам.

```python
import sys
from pathlib import Path

ROOT = Path.cwd()
if ROOT.name == "notebooks":
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT))
DATA = ROOT / "data" / "lab3"

from lab3.dataio import read_vector, read_scalar
from lab3.interpolation import (
    arccot, sample_function, lagrange_eval, newton_build, newton_eval,
    interpolation_error, poly_string_lagrange, poly_string_newton,
)

X_star = read_scalar(DATA / "3_1" / "X_star.txt")
nodes_a = read_vector(DATA / "3_1" / "nodes_a.txt")
nodes_b = read_vector(DATA / "3_1" / "nodes_b.txt")
print("X* =", X_star)
print("a) xi =", nodes_a)
print("b) xi =", nodes_b)
print("f(X*) =", arccot(X_star))
```

## a) Узлы −3, −1, 1, 3

```python

xs = nodes_a
ys = sample_function(xs)
print("yi =", ys)

print("Лагранж:")
print("L(x) =", poly_string_lagrange(xs, ys))
L_a = lagrange_eval(xs, ys, X_star)
print("L(X*) =", L_a)

coeffs, table = newton_build(xs, ys)
print("Ньютон:")
print("P(x) =", poly_string_newton(xs, coeffs))
N_a = newton_eval(xs, coeffs, X_star)
print("P(X*) =", N_a)

true_y = arccot(X_star)
err_a = interpolation_error(true_y, L_a)
print("|f(X*)-L(X*)| =", err_a)
print("|L-N| =", abs(L_a - N_a))
```

## b) Узлы −3, 0, 1, 3

```python

xs = nodes_b
ys = sample_function(xs)
print("yi =", ys)

print("Лагранж:")
print("L(x) =", poly_string_lagrange(xs, ys))
L_b = lagrange_eval(xs, ys, X_star)
print("L(X*) =", L_b)

coeffs, table = newton_build(xs, ys)
print("Ньютон:")
print("P(x) =", poly_string_newton(xs, coeffs))
N_b = newton_eval(xs, coeffs, X_star)
print("P(X*) =", N_b)

err_b = interpolation_error(true_y, L_b)
print("|f(X*)-L(X*)| =", err_b)
print("|L-N| =", abs(L_b - N_b))
```
