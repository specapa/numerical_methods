# 3.4. Численное дифференцирование

Реализация: `lab3/diff.py`. Первая (лево-/правосторонняя и O(h²)) и вторая производные в точке x*.

```python
import sys
from pathlib import Path

ROOT = Path.cwd()
if ROOT.name == "notebooks":
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT))
DATA = ROOT / "data" / "lab3"

from lab3.dataio import read_vector, read_scalar
from lab3.diff import differentiate_at

xs = read_vector(DATA / "3_4" / "x.txt")
ys = read_vector(DATA / "3_4" / "y.txt")
x_star = read_scalar(DATA / "3_4" / "x_star.txt")
print("x =", xs)
print("y =", ys)
print("x* =", x_star)

res = differentiate_at(xs, ys, x_star)
print("индекс узла:", res["index"])
print("y'_left  (O(h))  =", res["y_prime_left"])
print("y'_right (O(h))  =", res["y_prime_right"])
print("y'       (O(h^2)) =", res["y_prime"])
print("y''              =", res["y_double_prime"])
```
