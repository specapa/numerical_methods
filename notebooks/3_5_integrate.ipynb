# 3.5. Численное интегрирование

Реализация: `lab3/integrate.py`. Прямоугольники (средние), трапеции, Симпсон; уточнение Рунге–Ромберга.

```python
import sys
from pathlib import Path

ROOT = Path.cwd()
if ROOT.name == "notebooks":
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT))
DATA = ROOT / "data" / "lab3"

from lab3.dataio import read_vector
from lab3.integrate import integrate_all
import math

params = read_vector(DATA / "3_5" / "params.txt")
X0, Xk, h1, h2 = params
print("X0, Xk, h1, h2 =", X0, Xk, h1, h2)

def f(x):
    return 1.0 / (x * x + 64.0)

# точное значение: (1/8)(arctg(Xk/8) - arctg(X0/8))
exact = (math.atan(Xk / 8.0) - math.atan(X0 / 8.0)) / 8.0
print("exact =", exact)

results = integrate_all(f, X0, Xk, h1, h2)
for name, r in results.items():
    print("---", name, "p =", r["p"])
    print("  F(h1) =", r["h1"])
    print("  F(h2) =", r["h2"])
    print("  RR    =", r["runge_romberg"])
    print("  |R|   =", r["error_est"])
    print("  |RR-exact| =", abs(r["runge_romberg"] - exact))
```
